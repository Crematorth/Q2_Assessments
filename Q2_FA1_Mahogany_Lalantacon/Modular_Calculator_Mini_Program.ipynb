```python
# Input number variables.
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# List of valid operations
valid_operations = ["addition", "subtraction", "multiplication", "division"]

# Checks if the input is one of the valid operations.
while True:
    try:
        operation_used = input("Enter operation to be used: ")
        operation_used = str.lower(operation_used) # lowercases all letters
        operation_index = valid_operations.index(operation_used)
        break

    except ValueError:
        print("Invalid operation. Please try again.")


# Adds num1 and num2
def add_numbers(num1, num2):
  return num1 + num2

# Subtracts num1 and num2
def subtract_numbers(num1, num2):
 return num1 - num2

# Multiplies num1 and num2
def multiply_numbers(num1, num2):
  return num1 * num2

# Divides num1 and num2
def divide_numbers(num1, num2):
  if num2 == 0: # Checks if num2 is 0 to return an error.
    return "Cannot divide by zero."
  else:
   return num1 / num2


# States an empty variable
calculated_value = None

# This if-block finds the inputted operation and corresponds it with the function and returns it into the empty variable.
if operation_used == "addition":
  calculated_value = add_numbers(num1, num2)

elif operation_used == "subtraction":
  calculated_value = subtract_numbers(num1, num2)

elif operation_used == "multiplication":
  calculated_value = multiply_numbers(num1, num2)

elif operation_used == "division":
  calculated_value = divide_numbers(num1, num2)

# Displays the final calculated value.
print("The calculated value was: ",calculated_value)
```
